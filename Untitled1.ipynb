```python
# Libraries
import pandas as pd
import os
from sqlalchemy import create_engine


# Create SQLite engine
engine = create_engine('sqlite:///inventory.db')


# Function to load data into SQLite
def ingest_db(df, table_name, engine, first_chunk):
    
    df.to_sql(
        table_name,
        con=engine,
        if_exists='replace' if first_chunk else 'append',
        index=False
    )


# Read CSV files in chunks
for file in os.listdir('data'):
    
    # Process only CSV files
    if file.endswith('.csv'):
        
        print(f"\nProcessing: {file}")
        
        first_chunk = True
        total_rows = 0
        
        # Read 5,000 rows at a time
        for df in pd.read_csv(
            os.path.join('data', file),
            chunksize=5000
        ):
            
            # Get table name by removing .csv
            table_name = file[:-4]
            
            # Insert chunk into SQLite
            ingest_db(
                df,
                table_name,
                engine,
                first_chunk
            )
            
            # After first chunk, append remaining chunks
            first_chunk = False
            
            total_rows += len(df)
            
            print(f"Inserted: {total_rows:,} rows")
        
        print(f"Completed: {file}")
```
